```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
```

```python
for i in c.columns:
    if c[i].dtypes =='object':
        c[i]= le.fit_transform(c[i])
```

```python
c.head()
```

```python
ip=c.drop(["Exited","Surname","CustomerId","RowNumber"],axis=1)
op=c.Exited
```

```python
ip.head()
```

```python
op.head()
```

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest=train_test_split(ip,op,train_size=0.8)
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
```

```python
xtrain=sc.fit_transform(xtrain)
xtrain
```

```python
xtest=sc.transform(xtest)
xtest
```

```python
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression()
```

```python
lr.fit(xtrain,ytrain)
```

```python
pred = lr.predict(xtest)
pred
```

```python
ytest
```

```python
from sklearn.metrics import confusion_matrix,ConfusionMatrixDisplay
cm = confusion_matrix(ytest,pred)
cm
```

```python
cmd = ConfusionMatrixDisplay(cm)
cmd.plot()
plt.show()
```

```python
from sklearn.metrics import accuracy_score,recall_score,precision_score,f1_score
acc = accuracy_score(ytest,pred)
rec = recall_score(ytest,pred)
pre = precision_score(ytest,pred)
f1 = f1_score(ytest,pred)
```

```python
print(f"Accuracy:{acc}\nRecall:{rec}\nPrecision:{pre}\nF1 Score:{f1}")
```
