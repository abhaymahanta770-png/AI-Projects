```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
```

```python
for i in D.columns:
    if D[i].dtypes =='object':
        D[i]= le.fit_transform(D[i])
```

```python
ip = D.drop(['Unnamed: 0','price'],axis=1)
op = D.price
```

```python
ip.head()
```

```python
op.head()
```

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest=train_test_split(ip,op,train_size=0.8)
```

```python
from sklearn.preprocessing import StandardScaler
sc =StandardScaler()
```

```python
xtrain = sc.fit_transform(xtrain)
xtrain
```

```python
xtest = sc.transform(xtest)
xtest
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
```

```python
lr.fit(xtrain,ytrain)
```

```python
pred = lr.predict(xtest)
pred
```

```python
ytest
```

```python
from sklearn.metrics import mean_squared_error
mse = mean_squared_error(ytest,pred)
mse
```

```python
from sklearn.metrics import r2_score
r2= r2_score(ytest,pred)
r2
```

```python
df = pd.DataFrame({'Ytest':list(ytest),'Prediction':pred})
df.head()
```

```python
sns.lmplot(x='Ytest',y='Prediction',data=df,scatter_kws={'color':'m'},
           line_kws={'color':'r'})
plt.show()
```
