```python
ip=data2.drop(['Survived'],axis=1)
```

```python
op=data2['Survived']
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
ct=ColumnTransformer([('Sex',OneHotEncoder(),[1]),
                      ('Embarked',OneHotEncoder(),[6])],
                     remainder='passthrough')
ip=np.array(ct.fit_transform(ip),dtype=str)
```

```python
#train test split
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(ip,op,train_size=0.8)
```

```python
#standard scalar transform
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
x_train=sc.fit_transform(x_train)
x_test=sc.fit_transform(x_test)
```

```python
#Logistic regression
from sklearn.linear_model import LogisticRegression
alg1=LogisticRegression()
alg1.fit(x_train,y_train)
```

```python
yp=alg1.predict(x_test)
```

```python
yp
```

```python
df=pd.DataFrame({'y':list(y_test),'prediction':yp})
```

```python
df
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
cm=confusion_matrix(y_test,yp)
cm
```

```python
cm1=ConfusionMatrixDisplay(cm)
cm1.plot()
plt.show()
```

```python
from sklearn.metrics import accuracy_score,recall_score,precision_score,f1_score
acc=accuracy_score(y_test,yp)
recall=recall_score(y_test,yp)
precision=precision_score(y_test,yp)
f1=f1_score(y_test,yp)
```

```python
print(acc)
print(recall)
print(precision)
print(f1)
```
